# 【補足】保守インシデントの傾向分析と改善提案の回答例

対象データ：`保守インシデント分析_演習用_配布版.xlsx`（8,000件）

## 考え方の枠組み：What → Where → Why → How をピラミッドで組む

### 考える順と、伝える順は逆になる

| | 順序 | 形 |
|---|---|---|
| **考える** | What → Where → Why → How | ピラミッド（上から降りる。各層で候補を広げ、1つ選んで次へ） |
| **伝える** | 結論 → 根拠 → 事実 | 答えから話す（部門長の持ち時間は15分） |

以下は**考える順**のピラミッド。各層で**何を候補に挙げ、何を基準に選び、何を切ったか**がそのまま残る。

### まず問題を定義する — 問いは2つある

部門長は2つのことを言っている。**「件数は減らない」と「解決も遅い」は別の問題**で、探す場所も原因も打ち手も違う。それぞれ別のピラミッドになる。

#### 問い①　解決が遅い

| | 値 | 定義 |
|---|---|---|
| 現状 | **2,964件（37.1%）** | 8,000件中のSLA未達成 |
| 理想 | **1,760件（22.0%）** | 全件が「再割当なし」案件と同じ水準になった場合 |
| **ギャップ** | **1,204件（▲15.1pt）** | 受け渡しによって生まれている未達成 |

理想値を「未達成ゼロ」に置くと非現実的で、打ち手の大きさを測れない。**実績として存在する水準（再割当が起きなかった4,367件の22.0%）**を理想に据えると、ギャップがそのまま「受け渡しが生んでいる損失」になる。

ただし仮定を含む。再割当が起きなかった案件は、もともと簡単だった可能性がある。**理想値の置き方自体が分析者の判断**であり、そこを説明できることが条件になる。

#### 問い②　件数が減らない

こちらは理想値を置くより先に、**主張が本当かを確かめる。** 打ち切りの影響がない12週（2016-W09〜W20）の週次件数は傾き **−2.9件/週**。前半6週の平均628件に対し後半6週は597件で、ほぼ横ばい。**部門長の主張はデータに支持される。**

効果を定量化できないため、こちらはギャップを数字で置かない。**原因の特定までを答えとする。**

### 各層の選定基準

ピラミッドが正しく降りているかは、**各層で「なぜその枝を選んだか」が言えるか**で確かめる。基準を言葉にできない選択は、勘で選んだのと変わらない。

| 層 | 候補の出し方 | 選定基準 |
|---|---|---|
| **Where** | 手持ちの列で片端から層別する | **大きな層が実際に達成している水準**があり、かつ**操作できる**軸 |
| **Why** | 原因の仮説を並べる | **このデータで検証できる**もの |
| **How** | 打ち手を並べる | **効果を見積もれる**もの（見積もれないものは、そう明記して採る） |

**Where の基準が2つあるのが要点。** 拠点別は最良5.5%だが201件（2.5%）の小さな拠点で、「全拠点をこの水準に」は根拠にならない。カテゴリ別は最良20.8%で大きな層だが、**案件の性質なので変えられない。** 両方を満たすのは「受け渡しの有無」だけだった。

問い②でも同じ。拠点別は上位5種で68.1%と症状別（49.5%）より集中しているが、**拠点は「どこで起票されたか」であって「同じ事象か」を示さない。** 再発を止めるには、同じ事象の単位＝症状で見る必要がある。

**切った枝にも理由がある。** それを残すのがこのピラミッドの役割になる。

## 第1部　前処理でなされるべき4つの判断

この演習の採点は、集計の正しさよりも**前処理での判断**に重みがある。以下の4つは、気づかないまま進めると結論が変わる。

### 判断1　所要時間は「受付→解決」で測る。クローズ日時は使わない

解決からクローズまでの時間の**85.6%が119〜121時間**に収まっている（中央値120.55時間）。解決の5日後に自動クローズされる運用で、クローズ日時は担当者の仕事の速さを表していない。

クローズ日時で所要時間を出した答案は、所要時間の中央値が145.7時間になる。実際の21.7時間とは7倍違う。

### 判断2　解決日時が空欄の516件は、指標ごとに扱いを変える

| 指標 | 母数 | 理由 |
|---|---|---|
| 件数・SLA未達成率 | 8,000件（全件） | 除外すると未達成を取りこぼす |
| 解決所要時間 | 7,484件 | 未解決を0時間として扱えない |

**一律に除外した答案は未達成率を誤る。一律に残した答案は所要時間を誤る。** 目的ごとに母数を変え、それを明示できているかが分かれ目。

### 判断3　代表値は中央値を使う

中央値21.7時間に対し、平均183.4時間、最大7,316時間。平均値で語ると「解決に1週間かかる部門」に見えるが、実際は**51.8%が24時間以内**に解決している。

平均値は併記してよい。ただし主役にすると、打つべき対策を見誤る。

### 判断4　率で順位づけするときは母数を併記する

優先度「緊急」78件と「高」136件は**未達成率100%**。しかし合計214件は、全未達成2,964件の**7.2%**にすぎない。

率の順位と件数の順位は一致しない。どちらを採ったかを言えることが条件。

### 明細から見える特定の経路

`T_作業履歴`（45,401行）から担当グループの遷移を集計すると、上位に**双方向の往復**が現れる。

| From | To | 遷移件数 |
|---|---|---|
| Group 20 | Group 70 | 354 |
| Group 70 | Group 20 | 352 |
| Group 70 | Group 39 | 334 |
| Group 39 | Group 70 | 150 |

Group 20 と Group 70 のあいだを往復した回数が合計706回。ほぼ同数の双方向で、**どちらが担当すべきか決まっていない**ことを示す。

両方が関与した案件を切り出すと差は鮮明になる。

| 区分 | 件数 | SLA未達成率 | 解決時間 中央値 |
|---|---|---|---|
| Group 20 と 70 の両方が関与 | 543（6.8%） | 63.7% | 153.8h |
| それ以外 | 7,457 | 35.1% | 17.8h |

全体の6.8%が、未達成346件（全未達成の11.7%）を生んでいる。

### 一次受けの精度

| 初回担当グループ | 件数 | 再割当発生率 | SLA未達成率 | 解決時間 中央値 |
|---|---|---|---|---|
| Group 70 | 4,118 | 41.3% | 25.8% | 1.3h |
| Group 20 | 449 | 92.7% | 51.0% | 59.4h |
| Group 9 | 121 | 93.4% | 85.1% | **1,340.7h** |
| Group 64 | 203 | 1.0% | 14.3% | 0.1h |

**Group 9 が異常値。** 121件と少数だが、受け取った案件の93.4%を他へ回し、未達成率85.1%、解決まで中央値で約56日かかっている。

### 経過時間の内訳

明細から状態ごとの滞留時間を積み上げると、削減余地のある場所が分かる。

| 状態 | 滞留合計 | 全体比 | 滞留中央値 |
|---|---|---|---|
| Resolved | 1,211,832h | 47.0% | 120.6h |
| Active | 465,728h | 18.1% | 0.5h |
| **Awaiting User Info** | 381,856h | **14.8%** | 17.4h |
| New | 366,304h | 14.2% | 0.6h |
| Awaiting Vendor | 100,024h | 3.9% | 125.3h |

Resolved の47.0%は自動クローズ待ちの120時間で、短縮対象ではない。**実質的に削れる最大の滞留は Awaiting User Info。** 客待ちを経験した1,713件の未達成率は70.6%（未経験は27.9%）。

### 件数が減らない理由

ここまでは「遅さ」の話。**件数側は、2つの数字を見るだけで原因が分かる。**

**症状が極端に集中している。**

| 症状 | 件数 | 累積 | 問題管理に紐付く件数 |
|---|---|---|---|
| **Symptom 491** | **3,010** | 37.6% | 37件（1.2%） |
| Symptom 534 | 536 | 44.3% | 0件 |
| Symptom 116 | 162 | 46.4% | 1件 |
| Symptom 387 | 134 | 48.0% | 0件 |
| Symptom 4 | 115 | 49.5% | 1件 |

症状は331種類あるのに、**上位5種で全体の49.5%。Symptom 491 は単独で37.6%**を占める。

**にもかかわらず、問題管理に回っていない。**

問題管理に紐付くインシデントは全体で109件（**1.4%**）。残り98.6%は個別に対応して閉じている。同じ症状が3,000回起きているのに、根本原因の調査に一度も載っていない。

**インシデント管理は回っているが、問題管理が機能していない。** これが「件数が減らない」の直接的な答えになる。

ただし**効果は見積もれない。** Symptom 491 の根本原因を解消して何件減るかは、このデータからは分からない。原因の特定までが、このデータで言えることの限界になる。

## 第3部　重回帰分析

### 変数設計

| | 内容 |
|---|---|
| 目的変数 | 解決所要時間の対数 `=LN(1+時間)` |
| 対象行 | 7,484行（解決日時が空欄の516件を除外） |
| 説明変数A（受付時点） | 優先度レベル／影響度レベル／緊急度レベル／受付時刻／休日フラグ／営業時間外フラグ／電話受付フラグ |
| 説明変数B（対応過程） | 再割当回数／再オープン回数／更新回数／関与グループ数／ナレッジ利用／問題管理紐付け／変更要求紐付け／ベンダー関与 |

対数変換は必須。生の時間のままだと最大7,316時間の外れ値に引かれ、係数が解釈できなくなる。

### モデル比較
どれも性能が良くないので分析においてはあまり役立たない

| モデル | 説明変数 | R² | 解釈 |
|---|---|---|---|
| モデル1 | A のみ（7変数） | **0.020** | 受付票の情報だけでは所要時間はほぼ説明できない |
| モデル2 | A + B（15変数） | **0.475** | 精度は上がるが、追加分は対応後にしか分からない |
| モデル3 | モデル2 − 優先度レベル | 0.475 | ほぼ不変。優先度は他2変数と重複している |
| 参考A | モデル2 − 更新回数 | 0.278 | 1列抜くだけでR²が0.20落ちる |
| 参考B | モデル2 − プロセス計数4列 | 0.083 | 説明力の大半がここに依存 |

**説明力0.475のうち0.392（82%）が対応過程の変数に由来する。** 受付時点では取得できない値なので、このモデルは予測には使えない。

### モデル2の主な係数

対数変換しているため、係数は `EXP(係数)` で「何倍になるか」として読む。

| 変数 | 係数 | t値 | 時間倍率 |
|---|---|---|---|
| 更新回数 | 0.1352 | **52.94** | 1.14 |
| ナレッジ利用フラグ | 0.9783 | 17.35 | 2.66 |
| 関与グループ数 | 0.3474 | 8.97 | 1.42 |
| 再割当回数 | 0.1359 | 5.96 | 1.15 |
| 変更要求紐付けフラグ | 1.4280 | 5.67 | 4.17 |
| 電話受付フラグ | −0.9304 | −4.54 | 0.39 |

更新回数の t値 52.94 が突出している。これは「更新が多いほど時間がかかる」ではなく、**時間がかかった案件ほど更新が積み上がる**という逆方向の関係。目的変数とほぼ同義の変数を説明変数に入れてしまっている状態（リーケージ）。

### モデル1で現れる符号の逆転

| 変数 | 係数 | t値 |
|---|---|---|
| 優先度レベル | **+1.0995** | 3.82 |

優先度レベルは「緊急=4、低=1」で作っているので、**優先度が高いほど解決時間が長い**という関係になる。現場の通念と逆だが、優先度「緊急」「高」のSLA未達成率100%と整合している。

### 多重共線性

| 変数 | VIF |
|---|---|
| 優先度レベル | **13.11** |
| 緊急度レベル | 5.70 |
| 影響度レベル | 4.77 |
| その他の変数 | 1.00〜3.45 |

優先度は影響度×緊急度でほぼ機械的に決まる（優先度レベル ≒ 影響度 + 緊急度 − 2）。優先度レベルを外すと他の係数が動く。

| 変数 | モデル2 | モデル3 | 変化 |
|---|---|---|---|
| 影響度レベル | +0.1622 | −0.0063 | **符号反転** |
| 緊急度レベル | +0.8360 | +0.6442 | −23% |

1変数を抜くだけで係数が動く状態では、個々の係数をそのまま解釈できない。

### 相関と交絡：ナレッジ利用フラグ

単純な集計では「ナレッジを使うと遅い」と読める。しかし月で分けると様子が変わる。

| 条件 | 件数 | SLA未達成率 | 解決時間 平均 |
|---|---|---|---|
| 全期間 ナレッジあり | 1,184 | 74.8% | 284.3h |
| 全期間 ナレッジなし | 6,816 | 30.5% | 165.3h |
| **3月のみ** ナレッジあり | 1,102 | 78.3% | 284.0h |
| **3月のみ** ナレッジなし | 1,791 | 50.8% | 235.1h |
| **3月以外** ナレッジあり | 82 | 28.1% | 288.7h |
| **3月以外** ナレッジなし | 5,025 | 23.2% | 141.9h |

ナレッジ利用「あり」1,184件のうち**1,161件（98%）が2〜3月に集中**している。3月以外では未達成率の差が44.3ポイントから4.9ポイントへ縮む。

**ナレッジ利用フラグは「3月」の代理変数になっている。** 回帰でも、3月を除くと係数は0.978から0.655へ下がる。

## 結論

分析の順（What → Where → Why → How）ではなく、**報告の順（結論 → 根拠 → 施策）**で並べる。

### 2つの問いと、2つの答え


| 問い | 現状 | 答え |
|---|---|---|
| **① 解決が遅い** | SLA未達成 2,964件（37.1%）<br>理想1,760件（22.0%）との差＝**ギャップ1,204件** | **受け渡しを減らす**<br>2施策で227件を回収（ギャップの19%） |
| **② 件数が減らない** | 週次 −2.9件/週でほぼ横ばい<br>（主張はデータに支持される） | **再発を止める**<br>上位5症状を問題管理に載せる |


### 根拠

**見出し：なぜ「受け渡し」なのか**

| 根拠 | 数字 |
|---|---|
| **① 長期化は受け渡しで起きている** | 関与1グループ 未達26.1%／平均97.8h → 4以上 77.6%／475.8h<br>再割当なし 22.0%・中央値0.7h → あり 55.1%・87.0h |
| **② 受け渡しは2経路に集中している** | Group 20 ↔ 70 の往復706回（全遷移6,133回の11.5%）<br>両方が関与した543件（6.8%）が未達成346件を生む |
| **③ 件数が減らないのは再発を止めていないから** | 症状331種のうち上位5種で49.5%、Symptom 491 単独で37.6%<br>問題管理に紐付くのは全体の1.4%（109件）のみ |

①と②が「遅さ」、③が「件数」に対応する。

①の裏付けには**受付時の7変数では R²=0.020** という数字を添える。これが「受付時の仕分けを厳しくすればよいのでは」という反論を先に潰す。受付票を見ても所要時間は読めないので、プロセス側を変えるしかない。

### 問いの答え

**【遅さ】ギャップ1,204件のうち、どこまで埋まるか**

| 施策 | 根拠 | 回収 |
|---|---|---|
| **1. Group 20 ↔ 70 の担当範囲を文書で確定** | 両方関与543件は未達63.7%（それ以外35.1%） | **155件** |
| **2. Group 9 を一次受けから外す** | 初回121件の未達85.1%、中央値1,341h、再割当発生93.4% | **72件** |

**合計227件＝ギャップの19%。未達成率は37.1%→34.2%。** 残る81%は、個別の経路ではなく受け渡しの発生そのものを減らす次の一手になる。

**【件数】再発を止める**

| 施策 | 根拠 | 効果 |
|---|---|---|
| **3. 上位5症状を問題管理に載せる** | 上位5症状で全体の49.5%、問題管理の紐付けは1.4% | **見積もれない** |

効果を数字で置かないことを、**隠さずに言う。** Symptom 491 の根本原因を解消して何件減るかは、このデータからは分からない。分かるのは「再発防止に回っていない」という事実までである。

**無理に数字を作らないことが、この提案の誠実さになる。**

#### 提案1（最優先）　Group 20 と Group 70 の担当範囲を文書で確定する

| | |
|---|---|
| 根拠 | 両グループ間の往復706回。両方が関与した543件は未達成率63.7%・中央値153.8h（それ以外は35.1%・17.8h） |
| 期待効果 | 543件を全体平均35.1%まで下げられれば未達成が**約155件減**（全未達成の5.2%） |
| 障壁 | どちらが担当すべきか決まっていないのは、技術的に切り分けが難しい領域だから。線引きには両リーダーの合意と、迷う案件の棚卸しが必要 |

#### 提案2　Group 9 を一次受けから外す

| | |
|---|---|

### 欄外：この分析の限界

1. **相関であって因果ではない。** 「関与グループ数が多いと遅い」は「受け渡しをやめれば速くなる」の証明ではない。難しい案件だから複数グループをまたいだ、という逆向きの説明も同じくらい成り立つ
2. **R²0.475のうち8割は「対応した結果」の変数に由来する。** 予測には使えず、要因の当たりをつける道具として読むべき
3. **説明できていないばらつきが半分以上残る。** 担当者のスキル、障害の実際の内容、部品調達の待ちなど、記録されていない要因がある
4. **ナレッジ利用フラグは「3月」の代理変数。** 単純集計では「ナレッジを使うと遅い」と見えるが、3月以外では差がほぼ消える
5. **受付の約99%が2016年3〜5月。** 通年の季節性は判断できない。また3月だけ未達成率61.3%（4月24.1%、5月21.9%）で、この月に何が起きたかはデータからは分からない
