## 冒頭

> 「会員はこんなに増えたのに、なぜ1人あたりのお買い上げ額は増えないのだろうか」

2026年12月の夕方。食品スーパー「FreshMart（フレッシュマート）」本社の販促企画部で、Aさんは一枚の資料を見つめていた。会員数と、会員1人あたりの年間購入金額の、過去3年の推移を並べたものである。

会員数は3年で37万人から52万人に増えていた。アプリの入会キャンペーンは社内で「成功事例」とされている。その一方で、会員1人あたりの年間購入金額は8万2千円から8万4千円と、ほとんど動いていなかった。同じ期間に販促予算は10.2億円から12.0億円へと2割増えている。

Aさんは3カ月前に販促企画部に異動し、前任の主任から年間12億円の販促予算を引き継いだばかりだった。2月の経営会議に、来年度の配分計画をかけなければならない。

先週の会議で、営業本部長からこう言われた。

> 「会員を集めるのにこの数年でずいぶん金を使ってきた。でも、それが売上になっている実感がない。来年度の予算は総額を増やせない。その中で、どこにいくら使うべきかを、根拠をつけて持ってきてほしい」

オフィスにはもう誰も残っていない。Aさんは手元の資料を並べ直し、考え始めた。

## 食品スーパー業界

### 市場の停滞と他業態の侵食

国内の食品スーパー市場は、ここ数年約10兆円前後で推移している（添付資料1を参照）。人口減と世帯の小規模化が進む一方で、外食から内食への戻りが支えとなり、市場全体としては横ばいが続いている。

この横ばいの内側で、客の取り合いは激しくなっていた。最も大きな変化はドラッグストアである。医薬品と化粧品の高い利益率を原資に、飲料・加工食品・日用品をスーパーより安く売る店が増え、その食品売上は3年で2.6兆円から3.2兆円まで伸びた。ネットスーパーも年率10%を超えるペースで拡大している。

ただし、そのどちらも**生鮮食品だけは切り崩せていない**。青果・精肉・鮮魚は日々の相場が動き、店頭での加工と廃棄ロスの管理が必要になるため、参入が難しい。多くの食品スーパーが生鮮を最後の砦と見なしているのはこのためである。

### 会員アプリをめぐる競争

各社がここ数年注力してきたのが、会員アプリとID-POSの整備である。誰が何を買ったかを顧客単位で把握できれば、チラシのような一斉定額の販促から、個人ごとのクーポン配信に切り替えられる。販促費の効率が上がるという触れ込みで、業界全体が同じ方向に動いた。

その結果、入会特典の競争が起きた。「アプリ登録で即時500円引き」「初回クーポン3枚セット」といった訴求が並び、消費者は複数のスーパーのアプリを同時に持つようになっている。業界誌には「会員数は増えたが、会員の質を語る言葉を各社がまだ持っていない」という記事が載ったこともある。

## FreshMart社

### 沿革と店舗

FreshMartは1978年に東京都世田谷区で創業した食品スーパーである。2026年3月末時点で首都圏に48店舗を展開し、年商は約620億円。店舗は駅前型18店、住宅地型22店、郊外ロードサイド型8店。平均売場面積は約450坪。

業界内での位置づけは「生鮮のFreshMart」である。創業期から青果と精肉の仕入れに人をかけ、店舗の一番良い場所を生鮮売場に割いてきた。生鮮3部門（青果・精肉・鮮魚）の売上構成比は約30%で、業界平均を上回る。精肉と鮮魚の一部店舗では、対面販売（店員に量を伝えて切り分けてもらう形式）を残している。

### 商品構成

店舗の売場は8部門に分かれている。

| 区分 | 部門 |
| --- | --- |
| 生鮮 | 青果、精肉、鮮魚 |
| その他 | 惣菜、日配（牛乳・卵・豆腐など）、加工食品、菓子、日用品 |

社内では「生鮮を見てから店内を一周する客が良い客だ」という言い方が古くからあるが、これは現場の経験則として語られてきたもので、データで検証されたことはない。

### 競合環境

首都圏の同規模チェーンが3社、全国展開の大手が2社。価格帯も品揃えも大きな差はなく、商圏は重なっている。直近3年、競合の新規出店によってFreshMartの店舗が大きく売上を落とした例は報告されていない。

好立地への出店は既に一巡しており、今後の成長は既存店で作るしかないというのが経営の認識である。

## 会員制度と販促の仕組み

### 会員カードとアプリ

FreshMartの会員カードは2009年に導入され、2021年にスマホアプリに移行した。入会は無料で、レジでの提示またはアプリのバーコード読み取りでポイントがたまる。販売額に占める会員経由の比率は約73%で、この数年ほぼ変わっていない。

退会手続きはあるが、実際に手続きをする会員はほとんどいない。使わなくなった会員は、会員のまま来なくなる。

### 会員ランク

前年度の年間購入金額に応じて、3つのランクを毎年4月に決定している（添付資料2を参照）。ランクは年度途中では変わらない。

ゴールド会員にはポイント2倍に加えて、月に一度の限定クーポンとセールの優先案内がつく。この優遇が販促予算の最大の使い道になっている。

### クーポンの運用

クーポンはアプリで配信され、商品部がメーカーとの商談をもとに対象商品を決める。対象は加工食品・日用品・菓子が中心である（添付資料9を参照）。

生鮮3部門は対象になりにくい。仕入れ価格が日々変動するため、数週間先の値引き原資を事前に確定できず、クーポンの発行計画に乗せづらいというのが商品部の説明である。

### 直近3年の入会キャンペーン

2023年度から、アプリの新規入会を増やすキャンペーンを毎年実施している。入会時に500円分のポイントと、初回クーポン（加工食品・日用品が中心）3枚セットを渡すもので、レジでの声がけを全店で徹底した。

このキャンペーンは社内で高く評価されている。会員数は3年で1.4倍になり、目標を毎年上回った。店舗スタッフの表彰制度も設けられ、入会獲得数は店舗評価指標の一つになっている。

## Aさんのミッションと足元の業績

### ミッション

Aさんは2026年9月に店舗の副店長から本社販促企画部に異動した。ミッションは次年度の販促予算12億円の配分計画を作り、2月の経営会議で承認を得ることである。

制約は二つある。

1. 予算総額は増やせない。12億円の内訳を変えることしかできない
2. 店舗数は次年度も増えない。成長は既存店で作る

### 足元の業績

会員数は3年で1.4倍になったが、会員1人あたりの年間購入金額はほぼ横ばいのままである（添付資料3を参照）。会員売上の総額は伸びているが、その増分の大半は店舗数の増加と会員数の増加で説明できる。既存店ベースの売上はこの3年、年率＋0.8％〜＋1.2％で推移している。

部門別に見ると、伸びている部門とそうでない部門がはっきり分かれている（添付資料5）。

### 前任者からの引き継ぎ

前任の主任は販促企画部で4年間予算を扱い、社内での評価は高かった。引き継ぎ資料の要旨は添付資料10の通りである。報告書の内容に虚偽はなく、記載された数値はいずれも経理部の確認を経ている。

引き継ぎの最後に、前任はこう付け加えた。

> 「ゴールドの維持率は数字で追えるし、成果も出ている。ただ、予算の7割をそこに入れているのが正解なのかは、正直のところ検証できていない。他に良い配分があるなら、数字で示してほしい」

### Aさんが使えるもの

社内の定例レポート（添付資料1〜9）に加え、情報システム部に依頼して会員単位の購買データを抽出してもらった（添付資料13）。分析に使える時間は1週間である。

## 設問

**あなたがAさんだとして、次年度の販促予算12億円の配分案を作り、その根拠を説明しなさい。**

### 提出物

| | 内容 |
| --- | --- |
| ① | 会員1人あたり購入金額が伸びていない原因について、あなたの結論 |
| ② | その結論に至った分析の過程（集計表・グラフ・検定結果） |
| ③ | 投資すべき対象層の定義。誰が何人いて、その層を選んだ根拠は何か |
| ④ | 次年度の予算配分案（金額内訳付き）と、各施策の内容 |
| ⑤ | 施策の効果をどう測るか。測定の設計を含めて |

### 条件

- 予算総額は12億円。増額は認められない
- 対象層は、実際にシステムで抽出できる条件で定義すること。「優良なお客様」のような表現では配信対象を作れない
- 前任者の配分を維持する結論でも構わない。ただしその場合も、維持すべき根拠をデータで示すこと

### 進め方の注意

添付資料はすべて、FreshMartの社内で既に使われている定例レポートである。前任者も同じ資料を見てきた。同じ資料を同じ切り口で見る限り、同じ結論になる。

添付資料13の生データは、定例レポートにはない切り口を作るためにある。

## 添付資料 10〜13

### 資料10　前任者の年度報告書（要旨）

2025年度の販促施策は、以下の点で成果を上げたと考える。

- ゴールド会員の翌年度ランク維持率は94%で、過去5年で最も高い。優遇施策の効果と考えられる
- 新規入会者数は目標の118%。入会キャンペーンは3年連続で目標を達成している
- クーポンの利用率は21%で、業界平均（15〜18%）を上回る。対象商品の選定精度は高いと評価できる
- アプリの月次起動率は68%で前年より6ポイント上昇した
- 販促費の売上高比率は2.65%。前年の2.74%から改善している

注：本報告書の数値はいずれも経理部の確認を経ており、記述に誤りはない。
